```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

```

```python
data = [["Is Chatbot dev course available?",1],["Are there seats available for course?",1],["I want to enroll in Chatbot Development course",1],["Hi Good Morning",0],["Hello",0],["heya",0]]
```

```python
df = pd.DataFrame(data, columns = ['text', 'intent'])
```

```python
df
```

```python
x = df['text']
y = df['intent']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(x)
```

```python
from sklearn.svm import LinearSVC
clf = LinearSVC(max_iter = 800,C=0.1)
clf.fit(X,y)
```

```python
text_test = ["Are Chatbot seats available?"]
X_test = vectorizer.transform(text_test)
clf.predict(X_test)
```
