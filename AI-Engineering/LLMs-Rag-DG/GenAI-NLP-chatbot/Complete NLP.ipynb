### Tokenization
example: the cat sat on the mat.
tokenization:-
"the","cat","sat","on","the","mat","."

```python
import nltk
import re #regex
import heapq
import numpy as np
nltk.download('punkt')
```

```python
paragraph = "Paragraphs are the building blocks of papers. Many students define paragraphs in terms of length"
```

```python
sentences = nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python
words = nltk.word_tokenize(paragraph)
words
```

# Stemming
removes any last es or ies letters from words like
meeting -> meet
# Lemetization
looks on grammer

```python
nltk.download('wordnet')
```

```python
from nltk.stem import PorterStemmer,WordNetLemmatizer
```

```python
stemmer = PorterStemmer()
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words]
    sentences[i] = ' '.join(words)
```

```python
words
```

```python
sentences
```

```python
lematizer = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words= [lematizer.lemmatize(word) for word in words]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

# Stop words 
Example: am, the, etc.... 
it differenciates the noun, pronoun, verb,etc

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in stopwords.words('english')]
    sentences[i] = ' '.join(words)
```

```python
word_tags = []
for tw in tagged_words:
    word_tags.append(tw[0]+" - "+tw[1]+",")
```
